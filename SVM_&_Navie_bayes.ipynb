Question 1 :  What is Information Gain, and how is it used in Decision Trees?
- Information Gain is a measure used in decision trees to determine which feature best splits the data at each node. It is based on the concept of entropy, which measures the impurity or uncertainty in a dataset.

Information Gain calculates the reduction in entropy after a dataset is split on a particular feature.


Usage in Decision Trees:

- The feature with the highest Information Gain is selected for splitting.

- It helps build a tree that separates classes most effectively.

- Commonly used in ID3 and C4.5 algorithms.

Question 2: What is the difference between Gini Impurity and Entropy?
1. Gini Impurity

Gini Impurity measures the probability that a randomly chosen data point would be incorrectly classified if it were labeled according to the class distribution of the node.

Key Points:

- Value ranges from 0 to 0.5 for binary classification

- 0 indicates a pure node

- Computationally faster

- Commonly used in CART and Random Forest algorithms

2. Entropy

Entropy measures the amount of uncertainty or disorder in the data, based on information theory.

Key Points:

- Value ranges from 0 to 1 for binary classification

- 0 indicates a pure node

- Computationally slower due to logarithmic calculations

- Used in ID3 and C4.5 algorithms

3. Key Differences

Gini Impurity and Entropy differ mainly in how they measure impurity and in their computational complexity. Gini Impurity measures the probability that a randomly selected sample would be misclassified, while Entropy measures the amount of uncertainty or disorder in the data. Gini Impurity is simpler to compute because it does not involve logarithmic calculations, making it faster in practice. Entropy, on the other hand, is based on information theory and is more sensitive to changes in class distribution. Both are used to split nodes in decision trees and often produce similar results.

Question 3:What is Pre-Pruning in Decision Trees?

:- Pre-pruning is a technique used to stop a decision tree from growing too large during training. It prevents overfitting by limiting tree growth early.

Common Pre-Pruning Techniques:

- Limiting maximum tree depth (max_depth)

- Minimum samples required to split (min_samples_split)

- Minimum samples per leaf (min_samples_leaf)

- Setting a minimum impurity decrease

Advantages:

- Reduces overfitting

- Faster training

- Simpler and more interpretable trees

```python
# Question 4: Write a Python program to train a Decision Tree Classifier using Gini Impurity as the criterion and print the feature importances.

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# Load dataset
data = load_iris()
X = data.data
y = data.target

# Train Decision Tree using Gini Impurity
model = DecisionTreeClassifier(criterion='gini', random_state=42)
model.fit(X, y)

# Feature importances
feature_importances = pd.DataFrame({
    'Feature': data.feature_names,
    'Importance': model.feature_importances_
})

feature_importances
```

# Question 5: What is a Support Vector Machine (SVM)?

Support Vector Machine (SVM) is a supervised learning algorithm used for classification and regression tasks. It works by finding the optimal hyperplane that maximally separates different classes.

Key Concepts:

- Uses support vectors (data points closest to the hyperplane)

- Maximizes margin

- Effective in high-dimensional spaces

# Question 6:  What is the Kernel Trick in SVM?

The Kernel Trick allows SVMs to handle non-linearly separable data by transforming it into a higher-dimensional space where linear separation becomes possible.

Common Kernels:

- Linear

- Polynomial

- Radial Basis Function (RBF)

- Sigmoid

Benefit:
It avoids explicitly computing transformations, making SVM efficient.

```python
# Question 7:  Write a Python program to train two SVM classifiers with Linear and RBF kernels on the Wine dataset, then compare their accuracies.

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load dataset
wine = load_wine()
X = wine.data
y = wine.target

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Linear SVM
linear_svm = SVC(kernel='linear')
linear_svm.fit(X_train, y_train)
linear_pred = linear_svm.predict(X_test)

# RBF SVM
rbf_svm = SVC(kernel='rbf')
rbf_svm.fit(X_train, y_train)
rbf_pred = rbf_svm.predict(X_test)

# Accuracy
print("Linear SVM Accuracy:", accuracy_score(y_test, linear_pred))
print("RBF SVM Accuracy:", accuracy_score(y_test, rbf_pred))
```

# Question 8: What is the Naïve Bayes classifier, and why is it called "Naïve"?

Naïve Bayes is a probabilistic classifier based on Bayes’ Theorem. It is called “Naïve” because it assumes that all features are independent, which is rarely true in real-world data.

Advantages:

- Fast and efficient

- Works well with small datasets

- Performs well in text classification

# Question 9: Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes

Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes mainly differ in the type of data they assume and model.

- Gaussian Naïve Bayes assumes that features are continuous and follow a normal (Gaussian) distribution. It is commonly used for real-valued data such as measurements, heights, weights, or sensor values.

- Multinomial Naïve Bayes is designed for discrete count data. It is widely used in text classification, where features represent word frequencies or term counts in documents.

- Bernoulli Naïve Bayes works with binary features, where each feature indicates the presence or absence of something (e.g., whether a word appears in a document or not). It ignores how many times the feature occurs and only considers whether it occurs.

In short, Gaussian handles continuous data, Multinomial handles count-based data, and Bernoulli handles binary data.

```python
# Question 10:  Breast Cancer Dataset. Write a Python program to train a Gaussian Naïve Bayes classifier on the Breast Cancer dataset and evaluate accuracy.

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load dataset
data = load_breast_cancer()
X = data.data
y = data.target

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Train Gaussian Naive Bayes
model = GaussianNB()
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))
```
